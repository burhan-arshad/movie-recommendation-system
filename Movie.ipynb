```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
df=pd.read_csv('movies_metadata.csv')
df.head()
```

```python
df=df[['title', 'overview','genres','tagline','popularity','vote_average']]
```

```python
df=df.drop_duplicates().reset_index(drop=True)
```

```python
df.duplicated().sum()
```

```python
df=df.dropna(subset=['title'])
```

```python
df['overview']=df['overview'].fillna('')
```

```python
df['tagline']=df['tagline'].fillna('')
```

```python
df.isnull().sum()
```

```python
df.iloc[0]['genres']
```

```python
import ast
df['genres']=df['genres'].apply(lambda x: " ".join(i['name'] for i in ast.literal_eval(x)))
```

```python
df.head()
```

```python
df['tags']=df['overview']+" "+df['genres']+" "+df['tagline']
```

```python
df['tags'][0]
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import re
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
stop_words=set(stopwords.words('english'))
lemmatizer=WordNetLemmatizer()
```

```python
def preprocess_text(text):
  text=text.lower()
  text = re.sub(r'[^a-zA-Z\s]', '', text)
  words = text.split()
  words = [word for word in words if word not in stop_words]
  words = [lemmatizer.lemmatize(word) for word in words]
  return " ".join(words)
```

```python
df['tags']=df['tags'].apply(preprocess_text)
```

```python
df['tags'][1]
```

```python
df=df.reset_index(drop=True)
```

```python
indices=pd.Series(df.index,index=df['title']).drop_duplicates()
```

```python
indices
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer(stop_words='english', max_features=50000, ngram_range=(1,2))
tfidf_matrix=tfidf.fit_transform(df['tags'])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
def recommend_movies(title,n=10):
  if title not in indices:
    return["Moive Not Found"]
  idx=indices[title]
  sim_score=cosine_similarity(tfidf_matrix[idx],tfidf_matrix).flatten()
  sim_index=sim_score.argsort()[::-1][1:n+1]
  return df['title'].iloc[sim_index]
```

```python
recommend_movies('Toy Story',5)
```

```python
import pickle

pickle.dump(tfidf_matrix,open('tfidf_matrix.pkl','wb'))

pickle.dump(indices,open('indices.pkl','wb'))

df.to_pickle('df.pkl')

pickle.dump(tfidf,open('tfidf.pkl','wb'))
```
